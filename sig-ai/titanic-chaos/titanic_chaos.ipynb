```python
import pandas as pd

titanic=pd.read_csv('titanic.csv')
students = pd.read_csv('students.csv')

print("TITANIC")
print(titanic.info())
print(titanic.describe())
print(titanic.isnull().sum())
print(titanic.shape)

print("STUDENTS")
print(students.info())
print(students.describe())
print(students.isnull().sum())
print(students.shape)
```

Loaded both datasets and inspected their structures

```python
print(students['gender'].unique())
print(students['english_marks'].unique())
print(students.duplicated().sum())
```

```python
students['gender'] = students['gender'].str.lower()
students['gender'] = students['gender'].replace({'m': 'male', 'f': 'female'})
print(students['gender'].unique())
```

The gender column had 10 inconsistent variants (Male, M, m, MALE). Changed all values to lowercase, then mapped  m, f to their full form (male, female)

```python
students['english_marks'] = students['english_marks'].str.replace(' marks', '', regex=False)
```

```python
word_to_num = {
    'sixty': '60',
    'eighty': '80',
    'seventy five': '75'
}
students['english_marks'] = students['english_marks'].replace(word_to_num)
```

```python
students['english_marks'] = pd.to_numeric(students['english_marks'])
print(students['english_marks'].dtype)
print(students['english_marks'].unique())
```

The english_marks column mixed plain numbers, numbers with a " marks" suffix, and numbers spelled out as words (e.g. 'sixty'). Removed the suffix, mapped spelled-out words to digits, then converted the whole column to numeric

```python
print("Before:", students.shape)
students = students.drop_duplicates()
print("After:", students.shape)
print("Remaining duplicates:", students.duplicated().sum())
```

Found and removed 5 exact duplicate rows

```python
elderly_passengers = titanic.loc[titanic['Age'] > 60]
print(elderly_passengers)
print("Number of passengers over 60:", elderly_passengers.shape[0])
```

Used loc to filter passengers by a condition (Age > 60)

```python
subset = titanic.iloc[0:5, 0:4]
print(subset)
```

Used iloc to retrieve the first 5 rows and 4 columns by position, since iloc is for position-based selection regardless of column names

```python
titanic['Age'] = titanic.groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.median()))
print(titanic['Age'].isnull().sum())

for x in ['math_marks', 'physics_marks', 'chemistry_marks', 'english_marks']:
    students[x] = students[x].fillna(students[x].median())

print(students[['math_marks', 'physics_marks', 'chemistry_marks', 'english_marks']].isnull().sum())
```

Filled missing values in ageand mark column using the median Age within each Pclass group 

```python
pclass_summary = titanic.groupby('Pclass').agg({
    'Fare': 'mean',
    'Age': 'mean',
    'Survived': 'mean'
})
print(pclass_summary)
```

Grouped Titanic passengers by Pclass to compare mean Fare, mean Age, and survival rate

```python
port_lookup = pd.DataFrame({
    'Embarked': ['S', 'C', 'Q'],
    'Port_Name': ['Southampton', 'Cherbourg', 'Queenstown']
})

titanic = titanic.merge(port_lookup, on='Embarked', how='left')
print(titanic[['Embarked', 'Port_Name']].head(10))
```

Table mapping Embarked codes (S, C, Q) to their full port names (Southampton, Cherbourg, Queenstown), then merged it into the Titanic dataset
